# Value Iteration

The Bellman optimality backup replaces each state value with its best expected one-step return. The full transition model of a slippery grid is known.

## Known dynamics

There are four actions. The intended move occurs with probability 0.84 and either sideways slip has probability 0.08.

```python
import numpy as np
import matplotlib.pyplot as plt
from gallery.gridworld import (
    S,
    A,
    GAMMA,
    START,
    choose,
    step,
    q_from_v,
    exact_values,
    policy_report,
)
```

## Solve the grid

Entering the goal pays 1, entering the pit costs 1 and an ordinary step costs 0.04. The discount factor is 0.95.

```python
v = np.zeros(S)
residuals = []
for sweep in range(500):
    updated = q_from_v(v).max(axis=1)
    difference = abs(updated - v).max()
    residuals.append(difference)
    v = updated
    if difference < 1e-10:
        break
policy = q_from_v(v).argmax(axis=1)
print(
    f"Sweeps {len(residuals)} | Bellman optimality residual {abs(v-q_from_v(v).max(1)).max():.3g}"
)
```

## Evaluate the policy

Independent rollouts measure success and return. The numeric start value also has an exact solution.

```python
policy_report([("exact", policy)], [residuals], "Bellman residual", "Sweep")
plt.show()
```

## Reading the result

The heatmap contains discounted values. The simulated return is undiscounted and episodes stop at a terminal square or after sixty moves.
