# OPTICS

OPTICS builds a reachability ordering. A cut at a specified radius extracts the crescent-shaped groups from that ordering.

```python
from pathlib import Path
import sys

start = Path.cwd()
for parent in (start, *start.parents):
    if (parent / "gallery" / "__init__.py").is_file():
        sys.path.insert(0, str(parent))
        break
else:
    raise RuntimeError("Open this example inside the Machine Learning collection.")
```

## Fit the grouping

The generator supplies reference labels for scoring. The estimator sees only the coordinates.

```python
import matplotlib.pyplot as plt
from sklearn.cluster import OPTICS
from gallery.unsupervised import cluster_data, cluster_report
```

## Point cloud

```python
X, reference = cluster_data("moons")
model = OPTICS(min_samples=8, cluster_method="dbscan", eps=0.17)
labels = model.fit_predict(X)
```

## Group quality

External scores include every observation. Internal scores are calculated on assigned points.

```python
cluster_report(X, reference, labels)
import numpy as np

reach = model.reachability_[model.ordering_]
plt.figure(figsize=(6, 2.1))
plt.plot(np.where(np.isfinite(reach), reach, np.nan), linewidth=0.8)
plt.axhline(0.17, color="tab:red", linestyle="--")
plt.xlabel("OPTICS ordering")
plt.ylabel("Reachability")
plt.tight_layout()
plt.show()
```

## Reading the result

The adjusted Rand index compares fitted groups with the generating labels. The silhouette measures compactness, which can penalise a correctly recovered crescent or ring.
