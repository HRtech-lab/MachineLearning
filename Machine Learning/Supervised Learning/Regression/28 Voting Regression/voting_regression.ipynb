# Voting Regression

This ensemble averages a linear estimate with two tree-based estimates. Its weights are fixed and each component sees the same training data.

```python
from pathlib import Path
import sys

start = Path.cwd()
for parent in (start, *start.parents):
    if (parent / "gallery" / "__init__.py").is_file():
        sys.path.insert(0, str(parent))
        break
else:
    raise RuntimeError("Open this example inside the Machine Learning collection.")
```

## Estimator

The constructor below fixes the model used in every fold.

```python
import matplotlib.pyplot as plt
from sklearn.ensemble import HistGradientBoostingRegressor, RandomForestRegressor, VotingRegressor
from sklearn.linear_model import Ridge
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from gallery.tabular import concrete_split, regression_report

model = VotingRegressor(
    [
        ("ridge", make_pipeline(StandardScaler(), Ridge(alpha=3))),
        ("forest", RandomForestRegressor(n_estimators=70, min_samples_leaf=2, random_state=17)),
        ("boost", HistGradientBoostingRegressor(max_iter=90, random_state=17)),
    ]
)
```

## Data and split

All ages of a concrete recipe stay in the same partition. Five training folds also keep recipes together. The final test uses unseen mixes.

```python
X, X_test, y, y_test, groups = concrete_split()
```

## Held-out predictions

The table includes a training-only baseline. Both panels use the untouched test observations.

```python
predictions = regression_report(model, X, X_test, y, y_test, groups)
plt.show()
```

## Reading the result

Averaging helps when the components make different errors. A poorly matched component can also pull the estimate away from the target.
